```python
import pandas as pd
import requests

url="https://api.open-meteo.com/v1/forecast?latitude=40.5283&longitude=70.9414&current_weather=true"

response=requests.get(url)
data=response.json()

current=data["current_weather"]


print("--- Current Weather in Kokand ---")
print(f"Temperature: {current['temperature']} °C")
print(f"Wind Speed: {current['windspeed']} km/h")
print(f"Wind Direction: {current['winddirection']}°")

# Convert to Pandas DataFrame
df_weather = pd.DataFrame([current])
display(df_weather)
```
