```python
import pandas as pd
import requests

# Open Library API address for Python books
url = "https://openlibrary.org/search.json?q=python"

response = requests.get(url)
books_data = response.json()

books_list = []
for doc in books_data.get("docs", [])[:10]:
  title = doc.get("title", "N/A")
  author = doc.get("author_name", ["N/A"])[0]
  year = doc.get("first_publish_year", "N/A")

  books_list.append(
      {"Book Title": title, "Author": author, "First Publish Year": year}
  )

# Convert to Pandas DataFrame
df_books = pd.DataFrame(books_list)

print("--- Python Books List ---")
print(df_books)
```
