```python
import pandas as pd
import requests
```

```python
#api adress
url="https://open.er-api.com/v6/latest/USD"

response=requests.get(url)
data=response.json()

rates=data["rates"]

print("1 USD=", rates.get("UZS"), "UZS")
print("1 USD=", rates.get("EUR"), "EUR")
```

```python
#api adress
url="https://open.er-api.com/v6/latest/USD"

response=requests.get(url)
data=response.json()

rates=data["rates"]

#pandasga utkazish
df=pd.DataFrame(list(rates.items()), columns=["Valyuta", "Kurs_USD"])
uzs_rate=rates.get("UZS")
df['Kurs_UZS']=round(df['Kurs_USD']*uzs_rate, 2)

df.to_excel("valyuta_kurslari.xlsx", index=False)
df.to_csv('Valyuta_kuslari.csv',index=False)

print("Hammasi ok")
```
